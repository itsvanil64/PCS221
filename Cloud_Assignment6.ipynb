Q1. Word Count - Count the frequency of each word.
Input:
1) hadoop is fast
2) hadoop is scalable

```python
#manually
from collections import defaultdict

data = [
    "hadoop is fast",
    "hadoop is scalable"
]

# Mapper
mapped = []
for line in data:
    for word in line.split():
        mapped.append((word, 1))

# Shuffle
shuffled = defaultdict(list)
for key, value in mapped:
    shuffled[key].append(value)

# Reducer
result = {key: sum(values) for key, values in shuffled.items()}

print(result)
```

```python
#using mrjob
%%writefile q1_wordcount.py

from mrjob.job import MRJob

class WordCount(MRJob):

    def mapper(self, _, line):
        for word in line.split():
            yield word, 1

    def reducer(self, key, values):
        yield key, sum(values)

if __name__ == "__main__":
    WordCount.run()
```

Q2. Character Count - Count the frequency of each character (ignore spaces).
Input: big data

```python
#manually
from collections import defaultdict

text = "big data"

# Mapper
mapped = [(char, 1) for char in text if char != " "]

# Shuffle
shuffled = defaultdict(list)
for key, value in mapped:
    shuffled[key].append(value)

# Reducer
result = {key: sum(values) for key, values in shuffled.items()}

print(result)
```

Q3. Average Word Length (Per Word) - Compute the average length of each word.
Input: data science data big data

```python
#manually
from collections import defaultdict

text = "data science data big data"
words = text.split()

# Mapper
mapped = [(word, len(word)) for word in words]

# Shuffle
shuffled = defaultdict(list)
for key, value in mapped:
    shuffled[key].append(value)

# Reducer (average)
result = {word: sum(lengths)/len(lengths) for word, lengths in shuffled.items()}

print(result)
```

```python
#using mrjob
%%writefile q3_avgwordlen.py

from mrjob.job import MRJob

class AvgWordLen(MRJob):

    def mapper(self, _, line):
        for word in line.split():
            yield word, (len(word), 1)

    def reducer(self, key, values):
        total, count = 0, 0
        for l, c in values:
            total += l
            count += c
        yield key, total / count

if __name__ == "__main__":
    AvgWordLen.run()
```

Q6. Compute average marks for each student.
Input:
1) A 80
2) B 70
3) A 90
4) B 60
5) A 100

```python
#manually
from collections import defaultdict

data = [
    "A 80",
    "B 70",
    "A 90",
    "B 60",
    "A 100"
]

# Mapper
mapped = []
for line in data:
    student, marks = line.split()
    mapped.append((student, int(marks)))

# Shuffle
shuffled = defaultdict(list)
for k, v in mapped:
    shuffled[k].append(v)

# Reducer (average)
result = {k: sum(v)/len(v) for k, v in shuffled.items()}

print(result)
```

```python
#using mrjob
%%writefile q6_avg_marks.py

from mrjob.job import MRJob

class AvgMarks(MRJob):

    def mapper(self, _, line):
        student, marks = line.split()
        yield student, int(marks)

    def reducer(self, key, values):
        values = list(values)
        yield key, sum(values)/len(values)

if __name__ == "__main__":
    AvgMarks.run()
```

Q7. Compute average salary per department and Highest Paid Department (Based on Average
Salary)
Input:
1) HR 50000
2) IT 70000
3) HR 60000
4) IT 80000

```python
#mnually
data = [
    "HR 50000",
    "IT 70000",
    "HR 60000",
    "IT 80000"
]

# Mapper
mapped = []
for line in data:
    dept, salary = line.split()
    mapped.append((dept, int(salary)))

# Shuffle
shuffled = defaultdict(list)
for k, v in mapped:
    shuffled[k].append(v)

# Reducer (average)
avg_salary = {k: sum(v)/len(v) for k, v in shuffled.items()}

print("Average Salary:", avg_salary)

# Highest paid department
highest = max(avg_salary.items(), key=lambda x: x[1])
print("Highest Paid Department:", highest)
```

```python
%%writefile q7_salary.py

from mrjob.job import MRJob

class AvgSalary(MRJob):

    def mapper(self, _, line):
        dept, salary = line.split()
        yield dept, int(salary)

    def reducer(self, key, values):
        values = list(values)
        avg = sum(values)/len(values)
        yield key, avg

if __name__ == "__main__":
    AvgSalary.run()
```

Q8. Compute average temperature per country
1) New York,38
2) London,29
3) Tokyo,35
4) New York,32
5) Delhi,45
6) Ambala,35

```python
#manually
from collections import defaultdict

data = [
    "New York,38",
    "London,29",
    "Tokyo,35",
    "New York,32",
    "Delhi,45",
    "Ambala,35"
]

# 🔹 Mapper
mapped = []
for line in data:
    city, temp = line.split(",")
    mapped.append((city, int(temp)))

# 🔹 Shuffle
shuffled = defaultdict(list)
for k, v in mapped:
    shuffled[k].append(v)

# 🔹 Reducer (Average)
result = {k: sum(v)/len(v) for k, v in shuffled.items()}

print(result)
```

```python
%%writefile q8_temp_updated.py

from mrjob.job import MRJob

class AvgTemp(MRJob):

    def mapper(self, _, line):
        city, temp = line.split(",")
        yield city, int(temp)

    def reducer(self, key, values):
        values = list(values)
        yield key, sum(values)/len(values)

if __name__ == "__main__":
    AvgTemp.run()
```

```python
#load dataset--> the excel sheet
import pandas as pd

df = pd.read_csv("GlobalLandTemperatures_GlobalLandTemperaturesByMajorCity.csv")

df.head()
```

```python
#manually
from collections import defaultdict

# Remove missing values
df_clean = df.dropna(subset=["AverageTemperature"])

# Mapper
mapped = []
for _, row in df_clean.iterrows():
    country = row["Country"]
    temp = row["AverageTemperature"]
    mapped.append((country, temp))

# Shuffle
shuffled = defaultdict(list)
for k, v in mapped:
    shuffled[k].append(v)

# Reducer (Average)
result = {k: sum(v)/len(v) for k, v in shuffled.items()}

# Show first 10
print(list(result.items())[:10])
```

```python
#using mrjob
%%writefile q8_dataset.py

from mrjob.job import MRJob
import csv

class AvgTemp(MRJob):

    def mapper(self, _, line):
        try:
            row = next(csv.reader([line]))
            country = row[3]   # Country column
            temp = row[1]      # AverageTemperature column

            if temp != "":
                yield country, float(temp)
        except:
            pass

    def reducer(self, key, values):
        values = list(values)
        yield key, sum(values)/len(values)

if __name__ == "__main__":
    AvgTemp.run()
```
